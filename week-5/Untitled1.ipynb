```python
import numpy as np
# 1 np.unique()->return the unique elee=ment,removingduplicated values
# syntax
# np.unique(array)

arrays=[10,50,30,30,40]
arrays=np.array(arrays)
```

```python
# np.sort->function arranges array elements in ascending order by default.
# syntax
# np.sort(array)
# eg:  
np.sort(arrays)
```

```python
# np.argsort()->returns the indices that would arrange an array in sorted order.
# syntax
# np.argsort(array)
# eg:
np.argsort(arrays)
```

```python
# np.where()->select elements based on a condition and return their corresponding values.
# syntax
# np.where(condition, value_if_true, value_if_false)
# eg:
np.where(arrays>=40,"True","False")
```

```python
# 5. np.concatenate()
# What it does:The np.concatenate() function combines two or more arrays along an existing axis.
# Syntax:
# np.concatenate((array1, array2))
# eg:
a = np.array([10, 20, 30])
b = np.array([40, 50, 60])
np.concatenate((a,b))
```

```python
# np.clip()
# What it does:The np.clip() function limits array values to a specified minimum and maximum range.
# Syntax:
# np.clip(array,min,max)
# eg:
np.clip(arrays,10,30)
```

```python
# 7. np.argmax()
# What it does:The np.argmax() function returns the index of the maximum value in an array.
# Syntax:
# np.argmax(array)
np.argmax(arrays)
```

```python
# 8. np.argmin()
# What it does:The np.argmin() function returns the index of the minimum value in an array.
# Syntax:
# np.argmin()
# eg:
np.argmin(arrays)
```

```python
# 9. np.linspace()
# What it does:The np.linspace() function generates a specified number of evenly spaced values between a starting point and an ending point.
# Syntax:
# np.linespace(start,stop,num)
# eg:
np.linspace(0,20,5)
```

```python
# 10. np.eye()
# What it does:The np.eye() function creates a two-dimensional array with ones along its main diagonal and zeros elsewhere.
# Syntax:
# np.eye(N)
# eg:
np.eye(4)
```
