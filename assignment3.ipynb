```python
import pandas as pd
```

```python
data=pd.read_csv("dataset/LMS_Dirty_Dataset.csv")
# dataset exploration
# data.head(10)
# data.tail(10)
# data.shape
# data.columns
# data.dtypes
# data.describe()
```

```python
data.duplicated().sum()
data["student_id"].duplicated().sum()
```

```python
data.isnull().sum()
data["attendance_percent"].isnull()
data["attendance_percent"]=pd.to_numeric(data["attendance_percent"])
data["attendance_percent"]=data["attendance_percent"].fillna(data["attendance_percent"].median())
data.loc[25:35, ["attendance_percent"]]
```

```python
data.dtypes
data["enrollment_date"]=pd.to_datetime(data["enrollment_date"])
data.dtypes
```

```python
data["city"]=data["city"].str.title().str.strip()
data["course"]=data["course"].str.title().str.strip()
data["payment_status"] = data["payment_status"].str.title().str.strip()
data["completion_status"] = data["completion_status"].str.title().str.strip()
data
```

```python
data[data["course_fee"]<0]
data.loc[data["course_fee"]<0,"course_fee"]=0
data["course_fee"]=data["course_fee"].fillna(data["course_fee"])
data[data["rating"]<0]
```

```python
# course_fee
Q1=data["course_fee"].quantile(0.25)
print(Q1)
Q3=data["course_fee"].quantile(0.75)
print(Q3)
IQR=Q3-Q1
print(IQR)

lower_fee=Q1-1.5*IQR
print(lower_fee)
upper_fee=Q3+1.5*IQR
print(upper_fee)

outlier_fee=data[(data["course_fee"]<lower_fee)|(data["course_fee"]>upper_fee)]
outlier_fee


```

```python
#attendence_percnetage
Q1=data["attendance_percent"].quantile(0.25)
print(Q1)
Q3=data["attendance_percent"].quantile(0.75)
print(Q3)
iqr=Q3-Q1

lower_attendance=Q1-1.5*iqr
upper_attendance=Q3+1.5*iqr

outlier_attendeance=data[(data["attendance_percent"]<lower_attendance)|(data["attendance_percent"]>upper_attendance)]
outlier_attendeance
```

```python
q1=data["assignment_score"].quantile(0.25)
q3=data["assignment_score"].quantile(0.75)
iqr=q3-q1

lower_assignment=q1-1.5*iqr
upper_assignment=q3+1.5*iqr

outliers_assignment = data[(data["assignment_score"] < lower_assignment) |(data["assignment_score"] > upper_assignment)]

outliers_assignment
```

```python
q1=data["final_score"].quantile(0.25)
q3=data["final_score"].quantile(0.75)
iqr=q3-q1

lower=q1-1.5*iqr
upper=q3+1.5*iqr

outliers_final = data[(data["final_score"] < lower) |(data["final_score"] > upper)]

outliers_final
```

```python
data["course_fee"] = data["course_fee"].clip(lower_fee, upper_fee)
print(data["course_fee"])
data["assignment_score"]=data["assignment_score"].clip(lower_assignment,upper_assignment)
```

```python
mean = data["course_fee"].mean()
std = data["course_fee"].std()

z = (data["course_fee"] - mean) / std

outliers = data[(z > 3) | (z < -3)]

outliers
```

```python
# Attendance Percentage
mean = data["attendance_percent"].mean()
std = data["attendance_percent"].std()
z = (data["attendance_percent"] - mean) / std
outliers_attendance = data[(z > 3) | (z < -3)]
outliers_attendance
```

```python
# Assignment Score
mean = data["assignment_score"].mean()
std = data["assignment_score"].std()
z = (data["assignment_score"] - mean) / std
outliers_assignment = data[(z > 3) | (z < -3)]
outliers_assignment
```

```python
# Final Score
mean = data["final_score"].mean()
std = data["final_score"].std()
z = (data["final_score"] - mean) / std
outliers_final = data[(z > 3) | (z < -3)]
outliers_final
```
