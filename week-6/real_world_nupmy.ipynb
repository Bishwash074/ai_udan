```python
import pandas as pd
import numpy as np
```

```python
dd=pd.read_csv("daraz_dataset.csv")
dd.head()
dd.shape
dd.columns
dd.dtypes
dd["Age"]=pd.to_numeric(dd["Age"],errors="coerce")

numeric_columns=[
    "Age",
    "Quantity",
    "UnitPrice",
    "DiscountPercent",
    "TotalAmount",
    "Rating"
]
for columns in numeric_columns:
    dd[columns]=pd.to_numeric(dd[columns],errors="coerce")

    
dd.isnull().sum()
dd["Quantity"]=dd["Quantity"].fillna(dd["Quantity"].mean())
```

```python
features=["Age","Quantity","UnitPrice","DiscountPercent"]
target="TotalAmount"

# suppose features store garne X ,target store garne Y banau
X=dd[features]
y=dd[target]

type(X)
type(y)
# encoding or hartcoding if it is in string

X=X.to_numpy()
y=y.to_numpy()
type(X)
type(y)

X.shape
X.ndim
X.dtype
X.nbytes
X[0]
X[:,0]
X[:,3]
X[X[:,0]>30]

np.min(X[:,0])
```
