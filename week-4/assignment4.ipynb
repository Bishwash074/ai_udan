```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dd=pd.read_csv("dp_lms_student_analytics_dataset.csv")

```

```python
dd["EnrollmentDate"]=pd.to_datetime(dd["EnrollmentDate"])
dd["month"]=dd["EnrollmentDate"].dt.month
monthly_student=dd.groupby("month")["StudentID"].count()
print(monthly_student)
plt.figure(figsize=(12,6))
plt.plot(monthly_student.index,monthly_student.values)
plt.xlabel("Month",fontsize=20)
plt.ylabel("Number of student",fontsize=20)
plt.title("Monthly student Enrollment")
plt.grid(True)
```

```python
monthly_revenue=dd.groupby("month")["Revenue"].sum()
print(monthly_revenue)
plt.figure(figsize=(12,6))
plt.plot(monthly_revenue.index,monthly_revenue.values)
plt.xlabel("Month",fontsize=20)
plt.ylabel("Revenue",fontsize=20)
plt.grid(True)
```

```python
avg_courseprogress=dd.groupby("month")["CourseProgress"].mean()
plt.figure(figsize=(12,6))
plt.plot(avg_courseprogress.index,avg_courseprogress.values)
plt.title("Average Course Progress by Month")
plt.xlabel("Month")
plt.ylabel("Average Course Progress")
plt.grid(True)
```

```python
avgattendance_percentage=dd.groupby("month")["AttendancePercentage"].mean()
plt.figure(figsize=(12,6))
plt.plot(avgattendance_percentage.index,avgattendance_percentage.values)
plt.title("Average Attendance Percentage by Month")
plt.xlabel("Month")
plt.ylabel("Average Attendance (%)")
plt.grid(True)
```

```python
# 2Barchat
student_course=dd.groupby("Course")["StudentID"].count()
plt.figure(figsize=(12,6))
plt.bar(student_course.index,student_course.values,color="yellow")
plt.xlabel("Course")
plt.ylabel("Number of Students")
plt.title("Number of Students in Each Course")
plt.grid(True)
```

```python
revenue_course=dd.groupby("Course")["Revenue"].sum()
plt.figure(figsize=(12,6))
plt.bar(revenue_course.index,revenue_course.values,color="red")
plt.xlabel("Course")
plt.ylabel("Revenue")
plt.title("Revenue Generated by Each Course")
```

```python
final_score = dd.groupby("Course")["FinalScore"].mean()
plt.figure(figsize=(12,6))
plt.bar(final_score.index, final_score.values,color="green")

plt.xlabel("Course")
plt.ylabel("Average Final Score")
plt.title("Average Final Score by Course")
```

```python
attendance = dd.groupby("Course")["AttendancePercentage"].mean()

plt.figure(figsize=(12,6))
plt.bar(attendance.index, attendance.values,color="brown")

plt.xlabel("Course")
plt.ylabel("Average Attendance (%)")
plt.title("Average Attendance Percentage by Course")
```

```python
city = dd.groupby("City")["StudentID"].count()

plt.figure(figsize=(15,6))
plt.bar(city.index, city.values,color="brown")

plt.xlabel("City")
plt.ylabel("Number of Students")
plt.title("Number of Students from Each City")
```

```python
mentor = dd.groupby("Instructor")["StudentID"].count()

plt.figure(figsize=(12,6))
plt.bar(mentor.index, mentor.values,color="brown")

plt.xlabel("Mentor")
plt.ylabel("Number of Students")
plt.title("Number of Students Under Each Mentor")
```

```python
course=dd.groupby("Course")["StudentID"].count()
top10_course = course.sort_values(ascending=False).head(10)
plt.barh(top10_course.index, top10_course.values,color="brown")
plt.xlabel("Number of Students")
plt.ylabel("Course")
plt.title("Top 10 Courses by Enrollment")
```

```python
instructor=dd.groupby("Instructor")["StudentID"].count()
top10instructor=instructor.sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
plt.barh(top10instructor.index,top10instructor.values,color="brown")
plt.xlabel("Number of Students")
plt.ylabel("Mentor")
plt.title("Top 10 Mentors by Student Enrollment")
```

```python
highest_finalScore=dd.groupby("StudentName")["FinalScore"].count()
top10=highest_finalScore.sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
plt.barh(top10.index,top10.values,color="brown")
plt.xlabel("Number of Students")
plt.ylabel("Mentor")
plt.title("Top 10 Mentors by Student Enrollment")
```
