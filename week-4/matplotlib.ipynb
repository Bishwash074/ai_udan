```python
dd=pd.read_csv("LMS_Dirty_Dataset.csv")
```

```python
course=dd.groupby("course")["city"].count()
print(course)
plt.figure(figsize=(12,6))
plt.bar(course.index,course.values,color="green",edgecolor="blue")
plt.xlabel("course",fontsize=20)
plt.ylabel("city",fontsize=20)
plt.title("Course and City Bar graph")
plt.grid(axis="y",linestyle="--",alpha=1)

```
