```python
dd=pd.read_csv("LMS_Dirty_Dataset.csv")
```

```python
course=dd.groupby("course")["city"].count()
print(course)
plt.figure(figsize=(12,6))
plt.bar(course.index,course.values,color="green",edgecolor="blue")
plt.xlabel("course",fontsize=20)
plt.ylabel("city",fontsize=20)
plt.title("Course and City Bar graph")
plt.grid(axis="y",linestyle="--",alpha=1)

```

```python
course=dd.groupby("course")["city"].count()
print(course)
```

```python
# pie chart
colors=[
    "green",
    "red",
    "yellow",
    "orange",
    "purple",
    "pink",
    "brown"
]
plt.figure(figsize=(14,6))
plt.pie(course.values,labels=course.index,colors=colors,shadow=True,startangle=90,explode=(0.1,0,0,0,0,0,0,))
plt.title("City distrubution by city ")
```

```python
gender_course=dd.groupby("gender")["course"].count()
```

```python
colors=[
    "red",
    "yellow"
]
plt.pie(gender_course.values,colors=colors,labels=gender_course.index)
plt.title("Gender distribution by course")
```

```python
# histogram
# custom bin
bins=[0,30,50,80,100]
plt.hist(dd["assignment_score"],bins=bins,edgecolor="black",alpha=0.5)
plt.title("Distribution of assignment_score")
plt.xlabel("assignment_score")
plt.ylabel("Frequency")
```

```python
# scatter plot-relationship between two variable
data={
    "age":[24,25,26,27,28,30],
    "totalAmount":[1000,1200,1500,3000,3200,3400]
}
dfd=pd.DataFrame(data)
dfd
plt.scatter(dfd["age"],dfd["totalAmount"])
plt.xlabel("age")
plt.ylabel("totalAmount")
plt.grid(True)
plt.title("Age vs Total Amount")
```

```python
# scatter box
plt.scatter(dd["assignment_score"],dd["attendance_percent"])
plt.xlabel("assignmentscore")
plt.ylabel("attendancepercentage")
plt.grid(True)
plt.title("assinmentscore vs attendancepercentage")
```
